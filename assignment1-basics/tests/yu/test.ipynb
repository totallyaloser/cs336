```python
sample_str = """
low low low low low
lower lower widest widest widest
newest newest newest newest newest newest
"""
strings = sample_str.split()
my_dict = {}
for each_str in strings:
    if each_str not in my_dict:
        my_dict[each_str] = 1
    else:
        my_dict[each_str] = my_dict[each_str] + 1
vocab_counts: dict[tuple[bytes, ...], int] = {
    tuple(bytes([b]) for b in word.encode("utf-8")): count
    for word, count in my_dict.items()
}
vocab_counts
```

```python
word = "A"
encoded = word.encode("utf-8")  # b'A'

for b in encoded:
    print(f"b 的类型: {type(b)}, 值: {b}")
    # 输出: b 的类型: <class 'int'>, 值: 65

    # 1. bytes(b) —— 产生 65 个 \x00
    res1 = bytes(b)
    print(f"bytes(b): 长度={len(res1)}, 内容={res1}")
    # 输出: bytes(b): 长度=65, 内容=b'\x00\x00...\x00'

    # 2. bytes([b]) —— 包装成列表，构造单字节
    res2 = bytes([b])
    print(f"bytes([b]): 长度={len(res2)}, 内容={res2}")
    # 输出: bytes([b]): 长度=1, 内容=b'A'
```
